```python
# Cell 3 — Load BC5CDR from downloaded files
print("Loading BC5CDR dataset from local files...")

def read_conll_file(filepath):
    sentences = []
    tokens = []
    tags = []

    with open(filepath, 'r', encoding='utf-8') as f:
        for line in f:
            line = line.strip()
            if line == "" or line.startswith("-DOCSTART-"):
                if tokens:
                    sentences.append({
                        'tokens': tokens,
                        'tags': tags
                    })
                    tokens = []
                    tags = []
            else:
                parts = line.split('\t')
                if len(parts) >= 2:
                    tokens.append(parts[0])
                    tags.append(parts[-1])

    if tokens:
        sentences.append({'tokens': tokens, 'tags': tags})

    return sentences

# Load all splits
train_data = read_conll_file('data/train.tsv')
val_data   = read_conll_file('data/val.tsv')
test_data  = read_conll_file('data/test.tsv')

print(f"✅ Training samples:   {len(train_data)}")
print(f"✅ Validation samples: {len(val_data)}")
print(f"✅ Test samples:       {len(test_data)}")

print("\nFirst training example:")
print(train_data[0])

# Get all unique labels
all_tags = set()
for item in train_data + val_data + test_data:
    all_tags.update(item['tags'])

label_list = sorted(list(all_tags))
print(f"\nLabel list: {label_list}")

label2id = {l: i for i, l in enumerate(label_list)}
id2label  = {i: l for i, l in enumerate(label_list)}
print(f"\nLabel mapping: {label2id}")
```

```python
# Cell 3b — Convert to HuggingFace Dataset format
from datasets import Dataset, DatasetDict

def convert_to_hf_dataset(data, label2id):
    converted = {
        'tokens': [],
        'tags': []
    }
    for item in data:
        converted['tokens'].append(item['tokens'])
        converted['tags'].append([label2id[t] for t in item['tags']])
    return Dataset.from_dict(converted)

dataset = DatasetDict({
    'train':      convert_to_hf_dataset(train_data, label2id),
    'validation': convert_to_hf_dataset(val_data, label2id),
    'test':       convert_to_hf_dataset(test_data, label2id)
})

print("Dataset converted successfully:")
print(dataset)
print("\nFirst example:")
print(dataset['train'][0])
```

```python
# Cell 5 — Tokenization
label_list = dataset['train'].features['tags'].feature.names
label2id = {l: i for i, l in enumerate(label_list)}
id2label = {i: l for i, l in enumerate(label_list)}

def tokenize_and_align_labels(examples, tokenizer, max_length=128):
    tokenized = tokenizer(
        examples['tokens'],
        truncation=True,
        is_split_into_words=True,
        max_length=max_length,
        padding='max_length'
    )

    all_labels = []
    for i, tags in enumerate(examples['tags']):
        word_ids = tokenized.word_ids(batch_index=i)
        label_ids = []
        prev_word_id = None
        for word_id in word_ids:
            if word_id is None:
                label_ids.append(-100)
            elif word_id != prev_word_id:
                label_ids.append(tags[word_id])
            else:
                label_ids.append(-100)
            prev_word_id = word_id
        all_labels.append(label_ids)

    tokenized['labels'] = all_labels
    return tokenized

print("Tokenization function ready")
```

```python
# Cell 6 — Tokenization Function

def tokenize_and_align_labels(examples, tokenizer, max_length=128):
    tokenized = tokenizer(
        examples['tokens'],
        truncation=True,
        is_split_into_words=True,
        max_length=max_length,
        padding='max_length'
    )
    all_labels = []
    for i, tags in enumerate(examples['tags']):
        word_ids  = tokenized.word_ids(batch_index=i)
        label_ids = []
        prev_word_id = None
        for word_id in word_ids:
            if word_id is None:
                label_ids.append(-100)
            elif word_id != prev_word_id:
                label_ids.append(tags[word_id])
            else:
                label_ids.append(-100)
            prev_word_id = word_id
        all_labels.append(label_ids)
    tokenized['labels'] = all_labels
    return tokenized

print("✅ Tokenization function ready")
```

```python
# Cell 7 — Evaluation Function

def compute_metrics(eval_pred):
    predictions, labels = eval_pred
    predictions = np.argmax(predictions, axis=2)
    true_labels = [
        [id2label[l] for l in label_row if l != -100]
        for label_row in labels
    ]
    true_predictions = [
        [id2label[p] for p, l in zip(pred_row, label_row) if l != -100]
        for pred_row, label_row in zip(predictions, labels)
    ]
    return {
        "precision": precision_score(true_labels, true_predictions),
        "recall":    recall_score(true_labels, true_predictions),
        "f1":        f1_score(true_labels, true_predictions),
    }

print("✅ Evaluation function ready")
```

```python
# Cell 8 — Training Function

def train_model(model_name, display_name, epochs=3):
    print(f"\n{'='*60}")
    print(f"Training: {display_name}")
    print(f"Model:    {model_name}")
    print(f"{'='*60}")

    tokenizer = AutoTokenizer.from_pretrained(model_name)

    tokenized_dataset = dataset.map(
        lambda x: tokenize_and_align_labels(x, tokenizer),
        batched=True,
        remove_columns=dataset['train'].column_names
    )

    model = AutoModelForTokenClassification.from_pretrained(
        model_name,
        num_labels=len(label_list),
        id2label=id2label,
        label2id=label2id,
        ignore_mismatched_sizes=True
    )

    safe_name = display_name.replace(" ", "_")
    training_args = TrainingArguments(
        output_dir=f"results/models/{safe_name}",
        num_train_epochs=epochs,
        per_device_train_batch_size=16,
        per_device_eval_batch_size=16,
        learning_rate=2e-5,
        weight_decay=0.01,
        eval_strategy="epoch",
        save_strategy="epoch",
        load_best_model_at_end=True,
        metric_for_best_model="f1",
        logging_steps=50,
        fp16=torch.cuda.is_available(),
        report_to="none"
    )

    data_collator = DataCollatorForTokenClassification(tokenizer)

    trainer = Trainer(
        model=model,
        args=training_args,
        train_dataset=tokenized_dataset['train'],
        eval_dataset=tokenized_dataset['validation'],
        tokenizer=tokenizer,
        data_collator=data_collator,
        compute_metrics=compute_metrics
    )

    print(f"\nStarting training...")
    train_result = trainer.train()

    print(f"\nEvaluating on test set...")
    test_results = trainer.evaluate(tokenized_dataset['test'])

    # Detailed predictions
    predictions  = trainer.predict(tokenized_dataset['test'])
    preds        = np.argmax(predictions.predictions, axis=2)
    true_labels_list = [
        [id2label[l] for l in row if l != -100]
        for row in predictions.label_ids
    ]
    pred_labels_list = [
        [id2label[p] for p, l in zip(pr, lr) if l != -100]
        for pr, lr in zip(preds, predictions.label_ids)
    ]
    report = classification_report(
        true_labels_list, pred_labels_list, output_dict=True
    )

    results = {
        'model_name':            display_name,
        'precision':             test_results.get('eval_precision', 0),
        'recall':                test_results.get('eval_recall', 0),
        'f1':                    test_results.get('eval_f1', 0),
        'classification_report': report,
        'history':               trainer.state.log_history,
        'training_time':         train_result.metrics['train_runtime']
    }

    print(f"\n✅ {display_name} Results:")
    print(f"   Precision : {results['precision']:.4f}")
    print(f"   Recall    : {results['recall']:.4f}")
    print(f"   F1 Score  : {results['f1']:.4f}")

    return results, trainer, tokenizer

print("✅ Training function ready")
```

```python
# Cell 4 — Load and Parse Dataset
# Auto-reloads from disk if runtime restarted

def read_conll_file(filepath):
    sentences = []
    tokens = []
    tags = []
    with open(filepath, 'r', encoding='utf-8') as f:
        for line in f:
            line = line.strip()
            if line == "" or line.startswith("-DOCSTART-"):
                if tokens:
                    sentences.append({'tokens': tokens, 'tags': tags})
                    tokens = []
                    tags = []
            else:
                parts = line.split('\t')
                if len(parts) >= 2:
                    tokens.append(parts[0])
                    tags.append(parts[-1])
    if tokens:
        sentences.append({'tokens': tokens, 'tags': tags})
    return sentences

# Check if processed dataset already saved on disk
DATASET_SAVED = os.path.exists("data/bc5cdr_dataset") and \
                os.path.exists("data/label_info.json")

if DATASET_SAVED:
    print("✅ Processed dataset found on disk — loading...")
    dataset   = load_from_disk("data/bc5cdr_dataset")
    train_data = read_conll_file('data/train.tsv')
    val_data   = read_conll_file('data/val.tsv')
    test_data  = read_conll_file('data/test.tsv')

    with open("data/label_info.json", "r") as f:
        label_info = json.load(f)
    label_list = label_info['label_list']
    label2id   = label_info['label2id']
    id2label   = {int(k): v for k, v in label_info['id2label'].items()}
    print("✅ Dataset reloaded successfully")

else:
    print("Processing dataset for first time...")
    train_data = read_conll_file('data/train.tsv')
    val_data   = read_conll_file('data/val.tsv')
    test_data  = read_conll_file('data/test.tsv')

    # Build label list
    all_tags = set()
    for item in train_data + val_data + test_data:
        all_tags.update(item['tags'])
    label_list = sorted(list(all_tags))
    label2id   = {l: i for i, l in enumerate(label_list)}
    id2label   = {i: l for i, l in enumerate(label_list)}

    # Convert to HuggingFace Dataset
    def convert_to_hf(data):
        return Dataset.from_dict({
            'tokens': [d['tokens'] for d in data],
            'tags':   [[label2id[t] for t in d['tags']] for d in data]
        })

    dataset = DatasetDict({
        'train':      convert_to_hf(train_data),
        'validation': convert_to_hf(val_data),
        'test':       convert_to_hf(test_data)
    })

    # Save everything to disk
    dataset.save_to_disk("data/bc5cdr_dataset")
    with open("data/label_info.json", "w") as f:
        json.dump({
            "label_list": label_list,
            "label2id":   label2id,
            "id2label":   id2label
        }, f)
    print("✅ Dataset processed and saved to disk")

print(f"\n✅ Train:      {len(train_data):,} sentences")
print(f"✅ Validation: {len(val_data):,} sentences")
print(f"✅ Test:       {len(test_data):,} sentences")
print(f"✅ Labels:     {label_list}")
print(f"\nFirst example: {dataset['train'][0]}")
```

```python
# Cell 6 — Tokenization Function

def tokenize_and_align_labels(examples, tokenizer, max_length=128):
    tokenized = tokenizer(
        examples['tokens'],
        truncation=True,
        is_split_into_words=True,
        max_length=max_length,
        padding='max_length'
    )
    all_labels = []
    for i, tags in enumerate(examples['tags']):
        word_ids  = tokenized.word_ids(batch_index=i)
        label_ids = []
        prev_word_id = None
        for word_id in word_ids:
            if word_id is None:
                label_ids.append(-100)
            elif word_id != prev_word_id:
                label_ids.append(tags[word_id])
            else:
                label_ids.append(-100)
            prev_word_id = word_id
        all_labels.append(label_ids)
    tokenized['labels'] = all_labels
    return tokenized

print("✅ Tokenization function ready")
```

```python
# Cell 7 — Evaluation Function

def compute_metrics(eval_pred):
    predictions, labels = eval_pred
    predictions = np.argmax(predictions, axis=2)
    true_labels = [
        [id2label[l] for l in label_row if l != -100]
        for label_row in labels
    ]
    true_predictions = [
        [id2label[p] for p, l in zip(pred_row, label_row) if l != -100]
        for pred_row, label_row in zip(predictions, labels)
    ]
    return {
        "precision": precision_score(true_labels, true_predictions),
        "recall":    recall_score(true_labels, true_predictions),
        "f1":        f1_score(true_labels, true_predictions),
    }

print("✅ Evaluation function ready")
```
